## Challenge 01 - Revenue Analysis | Desafio 01 - Análise de Faturamento

```python
# Importing pandas. Matplotlib imported by pipenv install matplotlib

import pandas as pd
import matplotlib.pyplot as plt
```

### Dictionary creation | Criação do dicionário

```python
dict_faturamento = {
    'data_ref': [
        '2023-01-01', 
        '2020-02-01', 
        '2021-03-01', 
        '2022-04-01', 
        '2023-05-01',
        '2023-06-01', 
        '2020-07-01', 
        '2021-08-01', 
        '2022-09-01', 
        '2023-10-01',
        '2022-11-01', 
        '2023-12-01',
        ],
    'valor': [
        400000, 
        890000, 
        760000, 
        430000, 
        920000,
        340000, 
        800000, 
        500000, 
        200000, 
        900000,
        570000, 
        995000,
        ]
}
```

```python
dict_faturamento = pd.DataFrame.from_dict(dict_faturamento)
```

```python
dict_faturamento
```

### Average of revenue values | Média dos valores do faturamento

```python
dict_faturamento['valor'].mean()
```

### Vertical bar chart showing the reference month and value | Gráficos de barras vertical mostrando o mes de referência e o valor

```python
dict_faturamento.plot.bar(x='data_ref', y='valor')
```

### Line chart showing the reference month and value | Gráfico de barras de linhas mostrando o mês de referência e o valor

```python
dict_faturamento.plot.line(x='data_ref', y='valor')
```
